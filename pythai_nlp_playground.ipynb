```python
import timeit
from pythainlp.transliterate import romanize

# Sample words representing the different behaviors of 'ฤ'
test_words = ["ฤทธิ์", "ฤกษ์", "พฤษภาคม", "ทฤษฎี", "กฤษณ์", "กฤษณะ", "ฤดู"]

def check_sara_hardcoded(word: str) -> str:
    """Optimized Substring / Rule-based approach (Microseconds)"""
    if "ฤา" in word or "ฦา" in word:
        return "อือ"
    
    if "ฤกษ" in word:
        return "เออ"
    elif any(ex in word for ex in ("กฤษ", "ตฤณ", "ทฤษ", "ฤทธ", "ปฤษ", "ศฤง", "สฤต")):
        return "อิ"
    else:
        return "อึ"

def check_sara_romanized(word: str) -> str:
    """Heavy NLP Romanization approach (Milliseconds)"""
    if "ฤา" in word or "ฦา" in word:
        return "อือ"
        
    rom = romanize(word, engine="royin").lower()
    if "oek" in rom or "oe" in rom:
        return "เออ"
    elif "ri" in rom or "it" in rom:
        return "อิ"
    else:
        return "อึ"

# Wrapper functions for the benchmark loop
def run_hardcoded():
    for word in test_words:
        check_sara_hardcoded(word)

def run_romanized():
    for word in test_words:
        check_sara_romanized(word)

print("⚡ Starting Benchmark (Running 1,000 iterations per approach)...")
    
# Warm up pythainlp engine so initialization time isn't counted
print(romanize("พฤษภาคม", engine="royin"))

# Run the tests 1,000 times
iterations = 1000
time_hardcoded = timeit.timeit(run_hardcoded, number=iterations)

print("\n=== BENCHMARK RESULTS ===")
print(f"Hardcoded Rules: {time_hardcoded:.5f} seconds")
```

```python
from pythainlp.tokenize import word_tokenize, syllable_tokenize
from pythainlp.transliterate import pronunciate

POETRY_OVERRIDES = {
    "ก็": ["ก็"],
    "บ่": ["บ่"],
    "ได้": ["ได้"],
    "ธ": ["ทะ"],
    "ณ": ["นะ"],
    "ฤ": ["รึ"],
    "ฤๅ": ["รือ"],
    "ฤๅษี": ["รือ", "สี"],
    "มนุษย์": ["มะ", "นุด"],
    "พฤติกรรม": ["พฺรึด", "ติ", "กำ"],
    "สระ": ["สะ", "หระ"],
    "ความรู้สึก": ["ความ", "รู้", "สึก"],
    "เข้าใจ": ["เข้า", "ใจ"],
    "อยู่": ["อยู่"], # Fixes the "รู้อยู่" -> 'หยู่' hallucination
    "ผู้": ["ผู้"],   # Fixes the "ผู้ใด" -> 'พู่' hallucination
    "ใหม่": ["ใหม่"],
    "อย่า": ["อย่า"],
    "อยู่": ["อยู่"],
    "อย่าง": ["อย่าง"],
    "อยาก": ["อยาก"],
    "สุริยา": ["สุ", "ริ", "ยา"],
    "ก้ม": ["ก้ม"],
    "เดิน": ["เดิน"],
    "น้ำ": ["น้ำ"],
    "นำ": ["นำ"],
    "ทำ": ["ทำ"],
}
```

```python
def process_w2p(word: str) -> list:
    """Helper function to cleanly process a string through the w2p model."""
    # pronunciate hallucination on very short words. Use syllable_tokenize instead.
    if len(word) <= 2:
        return syllable_tokenize(word, engine="ssg")
        
    phonetic_word = pronunciate(word, engine="w2p")
    
    if not phonetic_word:
        return syllable_tokenize(word, engine="ssg")
    
    # Clean up unwanted characters like Phinthu (-ฺ) and hyphens (-)
    clean_phonetic = phonetic_word.replace("ฺ", "").replace("-", "")
    
    # Using ssg (CRF segmenter) as it handles phonetic text better than dict
    phonetic_syllables = syllable_tokenize(clean_phonetic, engine="ssg")
    
    # Clean up stray 'ห' artifacts
    return [s for s in phonetic_syllables if s != "ห" or word == "ห"]

def extract_poetic_syllables(text: str) -> list:
    """Extracts phonetic syllables for Klon 8 verification."""
    # Tokenize text into words using the newmm (greedy) engine, allowing w2p to handle compound words
    words = word_tokenize(text, engine="newmm")
    final_syllables = []
    
    for word in words:
        # Direct Override if the tokenized word is in the POETRY_OVERRIDES list (O(1) Fast Lookup)
        if word in POETRY_OVERRIDES:
            final_syllables.extend(POETRY_OVERRIDES[word])
            continue

        # Use 'ssg' to check if newmm greedy engine merged words like "ได้ใจ" or "รู้อยู่"
        sub_syllables = syllable_tokenize(word, engine="ssg")
        
        # Check if ANY of the segmented syllables are in the POETRY_OVERRIDES list
        has_override = any(sub in POETRY_OVERRIDES for sub in sub_syllables)
        
        if len(sub_syllables) > 1 and has_override:
            # Found a hidden override word in the segmented syllables. Process each sub-syllable independently.
            for sub in sub_syllables:
                if sub in POETRY_OVERRIDES:
                    final_syllables.extend(POETRY_OVERRIDES[sub])
                else:
                    final_syllables.extend(process_w2p(sub))
            continue
        
        # If no overrides were found inside, treat it as a true compound word (e.g. พัฒนาการ)
        # Proceed to parse the word into w2p pronunciate engine.
        final_syllables.extend(process_w2p(word))
        
    return final_syllables
```

```python
import csv
import io
from pythainlp.tokenize import word_tokenize, syllable_tokenize
from pythainlp.transliterate import pronunciate
from pythainlp.khavee import KhaveeVerifier

# Initialize Khavee for Rhyme Checking
kv = KhaveeVerifier()

def check_internal_klon8_rhymes(waks):
    """
    Goal 2: Automatically verify Klon 8 internal stanza rhymes using Khavee.
    Klon 8 Rule: 
    - Last syllable of Wak 1 rhymes with 3rd or 5th syllable of Wak 2
    - Last syllable of Wak 2 rhymes with Last syllable of Wak 3
    - Last syllable of Wak 3 rhymes with 3rd or 5th syllable of Wak 4
    """
    print("\n  [Rhyme Verification - Stanza 1 (Waks 1-4)]")
    
    # 1. Wak 1 End -> Wak 2 (3rd or 5th)
    w1_end = waks[1]["flat_syllables"][-1]
    w2_target_3 = waks[2]["flat_syllables"][2] if len(waks[2]["flat_syllables"]) > 2 else ""
    w2_target_5 = waks[2]["flat_syllables"][4] if len(waks[2]["flat_syllables"]) > 4 else ""
    
    match_w2 = kv.is_sumpus(w1_end, w2_target_3) or kv.is_sumpus(w1_end, w2_target_5)
    print(f"  - Wak 1->2 ({w1_end} -> {w2_target_3}/{w2_target_5}): {'✅ Pass' if match_w2 else '❌ Fail'}")

    # 2. Wak 2 End -> Wak 3 End
    w2_end = waks[2]["flat_syllables"][-1]
    w3_end = waks[3]["flat_syllables"][-1]
    match_w3 = kv.is_sumpus(w2_end, w3_end)
    print(f"  - Wak 2->3 ({w2_end} -> {w3_end}): {'✅ Pass' if match_w3 else '❌ Fail'}")

    # 3. Wak 3 End -> Wak 4 (3rd or 5th)
    w4_target_3 = waks[4]["flat_syllables"][2] if len(waks[4]["flat_syllables"]) > 2 else ""
    w4_target_5 = waks[4]["flat_syllables"][4] if len(waks[4]["flat_syllables"]) > 4 else ""
    match_w4 = kv.is_sumpus(w3_end, w4_target_3) or kv.is_sumpus(w3_end, w4_target_5)
    print(f"  - Wak 3->4 ({w3_end} -> {w4_target_3}/{w4_target_5}): {'✅ Pass' if match_w4 else '❌ Fail'}")


# ==========================================
# MAIN EXECUTION
# ==========================================

# csv_data = """w1_a,w1_b,w1_c,w2_a,w2_b,w2_c,w3_a,w3_b,w3_c,w4_a,w4_b,w4_c,w5_a,w5_b,w5_c,w6_a,w6_b,w6_c,w7_a,w7_b,w7_c,w8_a,w8_b,w8_c
# สมเด็จท้าว,บิตุรงค์,ดำรงราชย์,แสนสวาท,ลูกน้อย,เสน่หา,จะเสกสอง,ครองสมบัติ,ขัตติยา,แต่วิชา,สิ่งใด,ไม่ชำนาญ,จึงดำรัส,เรียกพระ,โอรสราช,มาริมอาสน์,แท่นสุวรรณ,แล้วบรรหาร,พ่อจะแจ้ง,เจ้าจงจำ,คำโบราณ,อันชายชาญ,เชื้อกษัตริย์,ขัตติยา
# แล้วก้มกราบ,บิตุราช,มาตุรงค์,ทั้งสององค์,ลูบหลัง,แล้วสั่งสอน,จะเดินทาง,กลางป่า,พนาดร,จงผันผ่อน,ตรึกจำ,คำโบราณ,จะพูดจา,สารพัด,บำหยัดยั้ง,จนลุกนั่ง,น้ำท่า,กระยาหาร,แม้นหลับนอน,ผ่อนพ้น,ที่ภัยพาล,อดบันดาล,โกรธขึ้ง,จึงสบาย"""
csv_path = "E:\\User\\Noto\\SIIT\\SIIT_Year_4\\LLM_WritePoem\\phraAphai_1_export.csv"

with open(csv_path, newline="", encoding="utf-8-sig") as f:
    reader = csv.DictReader(f)

    for row_idx, row in enumerate(reader):
        print(f"\n{'='*50}\nSTANZA {row_idx + 1}\n{'='*50}")
        
        # Dictionary to hold the structured waks for this row (1-indexed for clarity)
        waks_data = {}
        
        for i in range(1, 9):
            # Extract the 3 segments of the wak
            parts = [row[f'w{i}_a'], row[f'w{i}_b'], row[f'w{i}_c']]
            
            # Segment each part independently
            syllabled_parts = [extract_poetic_syllables(part) for part in parts]
            flat_syllables = [s for part_syls in syllabled_parts for s in part_syls]
            
            waks_data[i] = {
                "text_parts": parts,
                "syllabled_parts": syllabled_parts,
                "flat_syllables": flat_syllables
            }
            
            # Goal 1: Visualize the Segmentation
            text_display = " | ".join(parts)
            syl_display = " | ".join(["-".join(s) for s in syllabled_parts])
            
            print(f"Wak {i}: {text_display}")
            print(f"       [{syl_display}]")
            print("-" * 30)

        # Goal 2: Verify the rhymes
        check_internal_klon8_rhymes(waks_data)
```

```python
import timeit

setup_code = """
# Test words
word_match = "ธรรมชาติ"  # Ends with "ชาติ"
word_no_match = "หมูกรอบ" # Does not match any

# Original implementation (Set + Generator)
masking_frozenset = frozenset({
    "เกียรติ", "ชาติ", "ญาติ", "มัติ", "วัติ", "บัติ", "ญัติ",
    "ยัติ", "ภูมิ", "พฤติ", "พรรดิ", "วรรดิ", "พยาธิ", "โพธิ",
    "เกตุ", "เมรุ", "เหตุ", "ธาตุ", "วุฒิ", "สมมุติ", "วิมุติ",
})

# Optimized implementation (Tuple natively passed to endswith)
masking_tuple = (
    "เกียรติ", "ชาติ", "ญาติ", "มัติ", "วัติ", "บัติ", "ญัติ",
    "ยัติ", "ภูมิ", "พฤติ", "พรรดิ", "วรรดิ", "พยาธิ", "โพธิ",
    "เกตุ", "เมรุ", "เหตุ", "ธาตุ", "วุฒิ", "สมมุติ", "วิมุติ",
)
"""

# The test expressions
frozenset_match = "any(word_match.endswith(ex) for ex in masking_frozenset)"
tuple_match = "word_match.endswith(masking_tuple)"

frozenset_nomatch = "any(word_no_match.endswith(ex) for ex in masking_frozenset)"
tuple_nomatch = "word_no_match.endswith(masking_tuple)"

# Run benchmarks (1 million iterations each)
print("Testing Match ('ธรรมชาติ'):")
print(f"Frozenset: {timeit.timeit(frozenset_match, setup=setup_code, number=1000000):.4f} seconds")
print(f"Tuple:     {timeit.timeit(tuple_match, setup=setup_code, number=1000000):.4f} seconds")

print("\\nTesting No Match ('หมูกรอบ'):")
print(f"Frozenset: {timeit.timeit(frozenset_nomatch, setup=setup_code, number=1000000):.4f} seconds")
print(f"Tuple:     {timeit.timeit(tuple_nomatch, setup=setup_code, number=1000000):.4f} seconds")
```
